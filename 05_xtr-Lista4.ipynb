```python
# unificar todas as planilhas em uma estrutura de dados em painel. 
#  * Dicas: Lembre de adicionar uma coluna com o ano da amostra, 
#   * e uma coluna com um código identificador - Código do IBGE. 
#   Verifique o procedimento de criação do identificador que realizamos ao fazer o web scraping com o pandas em aula.

import pandas as pd

# Caminho do arquivo
file_path = r"C:\Users\johan\Downloads\arrecadacao-da-receita-administrada-pela-rfb-por-municipio-2005-a-2017.xlsx"

# Ler todas as abas disponíveis no arquivo
abas_disponiveis = pd.ExcelFile(file_path).sheet_names

# Criar um dicionário para armazenar os DataFrames com nomes ajustados
df_dict = {}

# Padronizar manualmente os nomes das abas
for aba in abas_disponiveis:
    aba_formatada = aba.strip()  # Remove espaços extras
    
    if aba_formatada.lower() == "gps":
        aba_padronizada = "GPS"
    elif aba_formatada.lower() == "darf":
        aba_padronizada = "DARF"
    elif aba_formatada.lower() == "total":
        aba_padronizada = "Total"
    else:
        aba_padronizada = aba_formatada  # Mantém outros nomes inalterados
    
    # Ler a aba com o nome original e salvar no dicionário com o nome padronizado
    df_dict[aba_padronizada] = pd.read_excel(file_path, sheet_name=aba, engine="openpyxl", skiprows=5)

# Exibir abas carregadas
df_dict.keys() # Confirma nomes ajustados
```

```python
# Criar DataFrame inicial vazio
list17 = None

# Aplicar melt() em todas as abas antes da unificação
for aba, df in df_dict.items():
    print(f"Length {aba}: {len(df)}")
    
    # Padronizar nome das colunas
    df.rename(columns={'MUNICIPIOS': 'Município', 'ARRECADAÇÃO': 'Arrecadação'}, inplace=True)

    # Transformar de wide para long
    df_melted = df.melt(id_vars=["Município", "UF"], var_name="Ano", value_name=f"Arrecadação_{aba}")

    # Gravar de onde vem o valor
    df_melted["Fonte"] = aba

    # Unificar corretamente usando merge
    if list17 is None:
        list17 = df_melted  # Primeiro DataFrame define a base
    else:
        list17 = list17.merge(df_melted, on=["Município", "UF", "Ano"], how="outer")

# Verificar os nomes das abas após a importação

print(list17.head())
```

```python
import pandas as pd

# Lista dos anos disponíveis
anos = [2018, 2019, 2020, 2021, 2022, 2023]

# Criar um dicionário para armazenar os DataFrames de cada ano
df_anos = {}

# Processar todos os arquivos em um loop
for ano in anos:
    print(f"📂 Processando {ano}...")

    # Definir o caminho do arquivo
    file_path = fr"C:\Users\johan\Downloads\arrecadacao-da-receita-administrada-pela-rfb-por-municipio-{ano}.xlsx"

    # Ler todas as abas disponíveis no arquivo
    abas_disponiveis = pd.ExcelFile(file_path).sheet_names

    # Criar um dicionário para armazenar os DataFrames de cada aba do ano
    df_dict = {}

    # Processar cada aba (DARF, GPS, Total)
    for aba in abas_disponiveis:
        aba_formatada = aba.strip()
        
        if aba_formatada.lower() == "gps":
            aba_padronizada = "GPS"
        elif aba_formatada.lower() == "darf":
            aba_padronizada = "DARF"
        elif aba_formatada.lower() == "total":
            aba_padronizada = "Total"
        else:
            aba_padronizada = aba_formatada  

        # Ler cada aba do arquivo e armazenar no dicionário
        df_dict[aba_padronizada] = pd.read_excel(file_path, sheet_name=aba, engine="openpyxl", skiprows=5)

    # Criar DataFrame consolidado para este ano
    df_final = None

    # Aplicar transformação `melt()` em todas as abas antes de armazenar
    for aba, df in df_dict.items():
        print(f"✅ Processando aba {aba} ({ano}): {len(df)} registros")

        # Padronizar nome das colunas
        df.rename(columns={'MUNICÍPIOS': 'Município', 'ARRECADAÇÃO': 'Arrecadação'}, inplace=True)

        # Adicionar a coluna 'Ano'
        df["Ano"] = ano

        # Transformar de wide para long
        df_melted = df.melt(id_vars=["Município", "UF", "Ano"], var_name="Categoria", value_name=f"Arrecadação_{aba}")

        # Gravar de onde vem o valor
        df_melted["Fonte"] = aba

        # Unificar corretamente usando merge
        if df_final is None:
            df_final = df_melted  # Primeiro DataFrame define a base
        else:
            df_final = df_final.merge(df_melted, on=["Município", "UF", "Ano", "Categoria"], how="outer")

    # Armazenar o resultado do ano no dicionário geral
    df_anos[ano] = df_final

    print(f"✅ Dados armazenados para {ano}!")

# Agora, `df_anos[2018]`, `df_anos[2019]`, etc. contêm os DataFrames de cada ano separadamente!

# visualizar
print(df_anos[2018].head())
print(df_anos[2019].head())
print(df_anos[2020].head())
print(df_anos[2021].head())
print(df_anos[2022].head())
print(df_anos[2023].head())
```

```python
import pandas as pd
from unidecode import unidecode

# Criar lista de municípios + UF a partir de list17 (2005-2017)
municipios_antigos = list17[["Município", "UF"]].drop_duplicates().dropna().reset_index(drop=True)

# Transformar em string e ordenar ignorando acentos
municipios_antigos["Município"] = municipios_antigos["Município"].astype(str)
municipios_antigos["UF"] = municipios_antigos["UF"].astype(str)
municipios_antigos = municipios_antigos.sort_values(by=["UF", "Município"], key=lambda x: x.apply(unidecode)).reset_index(drop=True)

# Criar lista de municípios + UF a partir de df_anos (2018-2023)
municipios_novos = pd.concat([df_anos[ano][["Município", "UF"]] for ano in df_anos], ignore_index=True).drop_duplicates().dropna().reset_index(drop=True)
municipios_novos["Município"] = municipios_novos["Município"].astype(str)
municipios_novos["UF"] = municipios_novos["UF"].astype(str)
municipios_novos = municipios_novos.sort_values(by=["UF", "Município"], key=lambda x: x.apply(unidecode)).reset_index(drop=True)

# Comparar listas de municípios
diferenca_municipios = municipios_novos.merge(municipios_antigos, on=["Município", "UF"], how="outer", indicator=True)
diferenca_municipios = diferenca_municipios[diferenca_municipios["_merge"] == "left_only"].drop(columns=["_merge"])

# Criar lista mestre combinando ambos os conjuntos
municipios_master = pd.concat([municipios_antigos, diferenca_municipios]).drop_duplicates().reset_index(drop=True)

# Exibir informações
print("✅ Municípios em list17 (2005-2017):", len(municipios_antigos))
print("✅ Municípios em df_anos (2018-2023):", len(municipios_novos))
print("✅ Diferença encontrada:", len(diferenca_municipios))
print("✅ Total de municípios na lista mestre:", len(municipios_master))
print(municipios_master.head())

```
