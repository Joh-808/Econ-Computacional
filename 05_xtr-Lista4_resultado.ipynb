```python
import pandas as pd

# Lista dos anos disponíveis
anos = [2018, 2019, 2020, 2021, 2022, 2023]

# Criar um dicionário para armazenar os DataFrames de cada ano
df_anos = {}

# Processar todos os arquivos em um loop
for ano in anos:
    print(f"📂 Processando {ano}...")

    # Definir o caminho do arquivo
    file_path = fr"C:\Users\johan\Downloads\arrecadacao-da-receita-administrada-pela-rfb-por-municipio-{ano}.xlsx"

    # Ler todas as abas disponíveis no arquivo
    abas_disponiveis = pd.ExcelFile(file_path).sheet_names

    # Criar um dicionário para armazenar os DataFrames de cada aba do ano
    df_dict = {}

    # Processar cada aba (DARF, GPS, Total)
    for aba in abas_disponiveis:
        aba_formatada = aba.strip()
        
        if aba_formatada.lower() == "gps":
            aba_padronizada = "GPS"
        elif aba_formatada.lower() == "darf":
            aba_padronizada = "DARF"
        elif aba_formatada.lower() == "total":
            aba_padronizada = "Total"
        else:
            aba_padronizada = aba_formatada  

        # Ler cada aba do arquivo e armazenar no dicionário
        df_dict[aba_padronizada] = pd.read_excel(file_path, sheet_name=aba, engine="openpyxl", skiprows=5)

    # Criar DataFrame consolidado para este ano
    df_final = None

    # Aplicar transformação `melt()` em todas as abas antes de armazenar
    for aba, df in df_dict.items():
        print(f"✅ Processando aba {aba} ({ano}): {len(df)} registros")

        # Padronizar nome das colunas
        df.rename(columns={'MUNICÍPIOS': 'Município', 'ARRECADAÇÃO': 'Arrecadação'}, inplace=True)

        # Adicionar a coluna 'Ano'
        df["Ano"] = ano

        # Transformar de wide para long
        df_melted = df.melt(id_vars=["Município", "UF", "Ano"], var_name="Categoria", value_name=f"Arrecadação_{aba}")

        # Gravar de onde vem o valor
        df_melted["Fonte"] = aba

        # Unificar corretamente usando merge
        if df_final is None:
            df_final = df_melted  # Primeiro DataFrame define a base
        else:
            df_final = df_final.merge(df_melted, on=["Município", "UF", "Ano", "Categoria"], how="outer")

    # Armazenar o resultado do ano no dicionário geral
    df_anos[ano] = df_final

    print(f"✅ Dados armazenados para {ano}!")

# Agora, `df_anos[2018]`, `df_anos[2019]`, etc. contêm os DataFrames de cada ano separadamente!

# visualizar
#print(df_anos[2018].head())
#print(df_anos[2019].head())
#print(df_anos[2020].head())
#print(df_anos[2021].head())
#print(df_anos[2022].head())
#print(df_anos[2023].head())
```

```python
from fuzzywuzzy import process

# Criar lista de municípios que não tiveram código IBGE
municipios_sem_ibge = municipios_master[municipios_master["Codigo_IBGE"].isna()]["Municipio_Padrao"].unique()

# Criar lista de municípios de códigos que não foram usados
municipios_nao_usados = codigos_nao_usados["Municipio_Padrao"].unique()

# Comparação fuzzy entre os dois conjuntos
correspondencias = {mun: process.extractOne(mun, municipios_nao_usados) for mun in municipios_sem_ibge}

# Exibir os municípios sem correspondência e sua melhor correspondência encontrada
print("🔍 Correspondências aproximadas:")
for mun, match in correspondencias.items():
    if match:  # Garante que houve uma correspondência
        print(f"{mun} → {match[0]} (similaridade: {match[1]}%)")
```

```python
print("🔍 Correspondências aproximadas:")
for municipio_original, (municipio_correspondente, similaridade) in correspondencias.items():
    # Identificar a UF correta do município original
    uf_original = municipios_master.loc[municipios_master["Municipio_Padrao"] == municipio_original, "UF"].values[0]

    # Identificar a UF correta do município correspondente
    uf_correspondente = codigos_nao_usados.loc[codigos_nao_usados["Municipio_Padrao"] == municipio_correspondente, "UF"].values
    uf_correspondente = uf_correspondente[0] if len(uf_correspondente) > 0 else "?"

    print(f"{municipio_original} ({uf_original}) → {municipio_correspondente} ({uf_correspondente}) (similaridade: {similaridade}%)")
```

```python
# Filtrar apenas correspondências com mais de 90% de similaridade
matches_validos = {orig: match for orig, match in correspondencias.items() if match[1] >= 90}

# Iterar pelos municípios e atualizar no DataFrame
for municipio_original, (municipio_correspondente, similaridade) in matches_validos.items():
    # Encontrar a linha correta no municipios_master
    index_mun = municipios_master.loc[municipios_master["Municipio_Padrao"] == municipio_original].index

    if not index_mun.empty:
        # Adicionar o nome alternativo
        municipios_master.loc[index_mun, "Nome_Alternativo"] = municipio_correspondente

        # Buscar e adicionar o código IBGE correto
        codigo_ibge = codigos_nao_usados.loc[codigos_nao_usados["Municipio_Padrao"] == municipio_correspondente, "Codigo_IBGE"].values
        if len(codigo_ibge) > 0:
            municipios_master.loc[index_mun, "Codigo_IBGE"] = codigo_ibge[0]

# Exibir resultado atualizado
print("🔍 Municípios atualizados:")
print(municipios_master[municipios_master["Municipio_Padrao"].isin(matches_validos.keys())])
```
